```python
import gradio as gr
```

```python
def greet(name, intensity):
  return "Hello, "+name+"!"*int(intensity)
```

```python
greet
```

```python
demo=gr.Interface(
    fn=greet,
    inputs=["text", "slider"],
    outputs=["text"],
    api_name="predict"
)
```

```python
demo.launch()
```
