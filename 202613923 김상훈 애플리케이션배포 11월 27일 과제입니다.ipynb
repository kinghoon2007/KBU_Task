```python
import seaborn as sns
import pandas as pd
import numpy as np
```

### <font color='00ff00'> 과제1: 데이터 탐색 및 전처리

```python
path = 'https://github.com/wesm/pydata-book/blob/3rd-edition/datasets/titanic/'
f'{path}train.csv'
```

```python
raw_path = 'https://raw.githubusercontent.com/wesm/pydata-book/3rd-edition/datasets/titanic/'
df_train = pd.read_csv(f'{raw_path}train.csv')
df_test = pd.read_csv(f'{raw_path}test.csv')
df_all = pd.concat([df_train, df_test], sort=True).reset_index(drop=True)
df_all
```

<font color='ffff00'> 1. 타이타닉 데이터의 구조와 주요 변수를 확인하시오.

```python
# 1. 행과 열의 크기 확인
print("Shape:", df_all.shape)

# 2. 열 목록, 비결측치 수, 데이터 타입 확인
print("\n--- Info ---")
df_all.info()

# 3. 결측치 개수 및 비율 확인
print("\n--- Missing Values ---")
missing = pd.DataFrame({
    "missing_count": df_all.isna().sum(),
    "missing_ratio": (df_all.isna().mean() * 100).round(2)
})
print(missing[missing["missing_count"] > 0])

# 4. 수치형 및 범주형 변수 요약 통계 확인
print("\n--- Numeric Summary ---")
print(df_all.describe())

print("\n--- Categorical Summary ---")
print(df_all.describe(include=["object", "string"]))

# 5. 상위/하위 샘플 확인
print("\n--- Head ---")
print(df_all.head())
```

<font color='ffff00'> 2. Age: Pclass와 Sex 그룹별 중앙값으로 대체하시오.

```python
df_all["Age"] = df_all.groupby(["Pclass", "Sex"])["Age"].transform(
    lambda s: s.fillna(s.median())
)
df_all
```

<font color='ffff00'> 3. Embarked: 최빈값으로 대체하시오.

```python
embarked_mode = df_all["Embarked"].mode(dropna=True)[0]
df_all["Embarked"] = df_all["Embarked"].fillna(embarked_mode)
df_all
```

<font color='ffff00'> 4. Fare: 중앙값으로 대체하시오.

```python
fare_median = df_all["Fare"].median()
df_all["Fare"] = df_all["Fare"].fillna(fare_median)
df_all
```

<font color='ffff00'> 5. Cabin: 결측 비율을 확인하고 컬럼 제거 또는 결측 여부 변수를 생성하시오.

```python
# Cabin 결측 비율 확인
cabin_missing_ratio = df_all["Cabin"].isna().mean()
print(f"Cabin 결측 비율: {cabin_missing_ratio:.2%}")

# 결측 여부 파생 변수(Cabin_Missing: 1=결측, 0=존재) 생성 후 원본 Cabin 컬럼 제거
df_all["Cabin_Missing"] = df_all["Cabin"].isna().astype(int)
df_all = df_all.drop(columns=["Cabin"])
```

<font color='ffff00'> 6. 불필요한 변수를 제거하시오. 이때, PassengerId, Name, Ticket 등 분석 목적에 따라 제거 여부를 결정하시오.

```python
cols_to_drop = ["PassengerId", "Name", "Ticket"]
df_all = df_all.drop(columns=[c for c in cols_to_drop if c in df_all.columns])

# 전처리 후 남은 결측치 확인
print(df_all.isna().sum())
```

### <font color='00ff00'> 과제2: Feature Engineering 및 데이터 변환

<font color='ffff00'> 1. Family_Size=SibSp+Parch+1을 하시오.

```python
df_all["Family_Size"] = df_all["SibSp"] + df_all["Parch"] + 1
df_all
```

<font color='ffff00'> 2. Family_Size를 기준으로 가족 규모를 Alone/Small/Medium/Large로 구분하시오.

```python
def categorize_family_size(size):
  if size == 1:
    return "Alone"
  elif size <= 3:
    return "Small"
  elif size <= 5:
    return "Medium"
  else:
    return "Large"


df_all["Family_Group"] = df_all["Family_Size"].apply(categorize_family_size)
df_all
```

<font color='ffff00'> 3. Name에서 승객의 호칭(Title)을 추출하시오.

```python
df_all["Name"] = pd.concat([df_train["Name"], df_test["Name"]], ignore_index=True)

# 9. Name에서 승객의 호칭(Title) 추출
df_all["Title"] = df_all["Name"].str.extract(r",\s*([^\.]+)\.", expand=False).str.strip()

title_mapping = {
    "Mlle": "Miss",
    "Ms": "Miss",
    "Mme": "Mrs",
    "Lady": "Rare",
    "Countess": "Rare",
    "Capt": "Rare",
    "Col": "Rare",
    "Don": "Rare",
    "Dr": "Rare",
    "Major": "Rare",
    "Rev": "Rare",
    "Sir": "Rare",
    "Jonkheer": "Rare",
    "Dona": "Rare",
}
df_all["Title"] = df_all["Title"].replace(title_mapping)

# 호칭 추출 완료 후 Name 컬럼 다시 제거
df_all = df_all.drop(columns=["Name"])
df_all
```

<font color='ffff00'> 4. 범주형 변수 인코딩을 하시오. 이때, Sex, Embarked, Pclass 등 One-Hot Encoding하시오.

```python
# 범주형 변수 One-Hot Encoding (타겟 변수 Survived 및 원본 Name 등 제외)
categorical_cols = [
    c
    for c in ["Sex", "Embarked", "Pclass", "Family_Group", "Title"]
    if c in df_all.columns
]

df_all = pd.get_dummies(
    df_all, columns=categorical_cols, drop_first=False, dtype=int
)
df_all
```

<font color='ffff00'> 5. StandardScaler를 사용하여 변수별 스케일 차이를 조정(수치형 변수 스케일링)하시오.

```python
from sklearn.preprocessing import StandardScaler

# 타겟 변수(Survived)와 원-핫 인코딩된 더미 변수(0/1)를 제외한 연속형/이산 수치형 변수 스케일링
numeric_cols = [
    c for c in ["Age", "Fare", "SibSp", "Parch", "Family_Size"] if c in df_all.columns
]

scaler = StandardScaler()
df_all[numeric_cols] = scaler.fit_transform(df_all[numeric_cols])
df_all
```

### <font color='00ff00'> 과제3: KNN 모델 학습 및 교차검

<font color='ffff00'> 1. 입력 변수(X)와 정답 변수(Y)로 분리하시오. 이때, X는 전처리 및 특성 변환이 완료된 입력 데이터이고, Y는 Survived이다.

```python
# df_all은 train과 test가 합쳐진 상태이므로 Survived가 결측치(NaN)가 아닌 학습 데이터만 분리
df_train_proc = df_all[df_all["Survived"].notna()].copy()

X = df_train_proc.drop(columns=["Survived"])
Y = df_train_proc["Survived"].astype(int)

print("X shape:", X.shape)
print("Y shape:", Y.shape)
```

<font color='ffff00'> 2. K-Nearest Neighbor 분류 모델을 생성하시오. 이때, sklearn.neighbors.KNeighborsClassifier를 사용하시오.

```python
from sklearn.neighbors import KNeighborsClassifier

# 기본 이웃 수(n_neighbors=5)로 KNN 분류 모델 생성
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model
```

<font color='ffff00'> 3. StratifiedKFold를 활용한 5-Fold 교차검증을 수행하시오. 단, 각 Fold에서 생존자와 비생존자의 비율을 유지하시오.

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

# 각 Fold에서 생존(1)과 비생존(0) 비율을 유지하는 Stratified 5-Fold 설정
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_scores = cross_val_score(
    knn_model, X, Y, cv=skf, scoring="accuracy"
)
cv_scores
```

<font color='ffff00'> 4. Accuracy를 계산하고 결과를 확인하시오.

```python
# Fold별 Accuracy 및 평균/표준편차 확인
for fold_idx, score in enumerate(cv_scores, start=1):
  print(f"Fold {fold_idx} Accuracy: {score:.4f}")

print(f"\nMean Accuracy: {cv_scores.mean():.4f}")
print(f"Std Accuracy:  {cv_scores.std():.4f}")
```

### <font color='00ff00'> 과제4: 하이퍼파라미터 튜닝 및 성능 평가

<font color='ffff00'> 1. 5-Fold Stratified Cross Validation을 설정하시오.

```python
from sklearn.model_selection import StratifiedKFold

# 각 Fold에서 생존(0/1) 클래스 비율을 유지하는 5-Fold 교차검증 분할기 설정
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
skf
```

<font color='ffff00'> 2. GridSearchCV를 이용한 최적 파라미터를 탐색하시오.

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder, StandardScaler


# 1) 파이프라인 내부에서 실행할 특성 엔지니어링 함수 (Fold별 독립 변환)
def engineer_features(df):
  df = df.copy()
  df["Family_Size"] = df["SibSp"] + df["Parch"] + 1

  df["Family_Group"] = np.select(
      [
          df["Family_Size"] == 1,
          df["Family_Size"] <= 3,
          df["Family_Size"] <= 5,
      ],
      ["Alone", "Small", "Medium"],
      default="Large",
  )

  if "Name" in df.columns:
    title = (
        df["Name"].str.extract(r",\s*([^\.]+)\.", expand=False).str.strip()
    )
    title_mapping = {
        "Mlle": "Miss",
        "Ms": "Miss",
        "Mme": "Mrs",
        "Lady": "Rare",
        "Countess": "Rare",
        "Capt": "Rare",
        "Col": "Rare",
        "Don": "Rare",
        "Dr": "Rare",
        "Major": "Rare",
        "Rev": "Rare",
        "Sir": "Rare",
        "Jonkheer": "Rare",
        "Dona": "Rare",
    }
    df["Title"] = title.replace(title_mapping).fillna("Rare")

  if "Cabin" in df.columns:
    df["Cabin_Missing"] = df["Cabin"].isna().astype(int)

  drop_cols = [
      c
      for c in ["PassengerId", "Name", "Ticket", "Cabin", "Survived"]
      if c in df.columns
  ]
  return df.drop(columns=drop_cols)


num_features = ["Age", "Fare", "SibSp", "Parch", "Family_Size", "Cabin_Missing"]
cat_features = ["Sex", "Embarked", "Pclass", "Family_Group", "Title"]

# 2) 수치형/범주형 전처리 파이프라인 (각 Fold의 train 데이터로만 fit 수행)
num_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

cat_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_transformer, num_features),
        ("cat", cat_transformer, cat_features),
    ],
    remainder="drop",
)

# 3) 전체 모델 파이프라인 구성
knn_pipeline = Pipeline(steps=[
    ("feat_eng", FunctionTransformer(engineer_features, validate=False)),
    ("preprocess", preprocessor),
    ("knn", KNeighborsClassifier()),
])

# 4) 탐색할 하이퍼파라미터 그리드 설정
param_grid = {
    "preprocess__num__imputer__strategy": ["median", "mean"],
    "knn__n_neighbors": [3, 5, 7, 9, 11, 15, 21],
    "knn__weights": ["uniform", "distance"],
    "knn__metric": ["euclidean", "manhattan", "minkowski"],
    "knn__p": [1, 2],
}

# 5) GridSearchCV 설정 및 학습
X_raw = df_train.drop(columns=["Survived"]).copy()
y_raw = df_train["Survived"].astype(int)

grid_search = GridSearchCV(
    estimator=knn_pipeline,
    param_grid=param_grid,
    cv=skf,
    scoring="accuracy",
    n_jobs=-1,
    refit=True,
)
grid_search.fit(X_raw, y_raw)
```

<font color='ffff00'> 3. 최적 하이퍼파라미터 및 교차 검증 성능(Accuracy)를 확인하시오.

```python
print("최적 하이퍼파라미터:")
for param, val in grid_search.best_params_.items():
  print(f"  - {param}: {val}")

print(f"\n최적 모델 5-Fold CV 평균 Accuracy: {grid_search.best_score_:.4f}")

# 상위 5개 파라미터 조합 결과 확인
cv_results_df = pd.DataFrame(grid_search.cv_results_).sort_values(
    by="rank_test_score"
)
cols_to_show = [
    "rank_test_score",
    "mean_test_score",
    "std_test_score",
    "param_knn__n_neighbors",
    "param_knn__weights",
    "param_knn__metric",
]
print("\n상위 5개 조합:")
print(cv_results_df[cols_to_show].head(5).to_string(index=False))
```

<font color='ffff00'> 4. 튜닝 전후 모델의 성능(Accuracy)를 비교하시오.

```python
from sklearn.model_selection import cross_val_score

# 동일 전처리 파이프라인 + 기본 KNN(n_neighbors=5, uniform, minkowski/p=2) 기준선 모델
baseline_pipeline = Pipeline(steps=[
    ("feat_eng", FunctionTransformer(engineer_features, validate=False)),
    ("preprocess", preprocessor),
    ("knn", KNeighborsClassifier(n_neighbors=5)),
])

baseline_scores = cross_val_score(
    baseline_pipeline, X_raw, y_raw, cv=skf, scoring="accuracy", n_jobs=-1
)
tuned_scores = cross_val_score(
    grid_search.best_estimator_,
    X_raw,
    y_raw,
    cv=skf,
    scoring="accuracy",
    n_jobs=-1,
)

base_mean, base_std = baseline_scores.mean(), baseline_scores.std()
tuned_mean, tuned_std = tuned_scores.mean(), tuned_scores.std()
diff = tuned_mean - base_mean

comparison_df = pd.DataFrame({
    "모델": ["튜닝 전 (기본 KNN)", "튜닝 후 (최적 KNN)"],
    "CV 평균 Accuracy": [round(base_mean, 4), round(tuned_mean, 4)],
    "CV 표준편차(Std)": [round(base_std, 4), round(tuned_std, 4)],
    "Fold별 Accuracy": [
        [round(s, 4) for s in baseline_scores],
        [round(s, 4) for s in tuned_scores],
    ],
})

print(comparison_df.to_string(index=False))
print(
    f"\n성능 변화량 (튜닝 후 - 튜닝 전): {diff:+.4f} ({diff * 100:+.2f}%p)"
)
```
