## <font color="00FF00"> 9. 교차검증과 하이퍼파라미터 튜닝

```python
mglearn.plots.plot_cross_validation()
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier

param_grid = {'n_neighbors': list(range(1, 16))}

knn = KNeighborsClassifier()

grid_search = GridSearchCV(knn, param_grid, cv=5, return_train_score=True)

grid_search.fit(X_train, Y_train)

print(f"최적의 하이퍼파라미터: {grid_search.best_params_}")
print(f"최고 교차 검증 정확도: {grid_search.best_score_:.4f}")
```

```python
best_model=grid_search.best_estimator_
test_accuracy=best_model.score(X_test, Y_test)
print(f"테스트 세트 정확도: {test_accuracy:.4f}")
```
